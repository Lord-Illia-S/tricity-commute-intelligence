```python
from pathlib import Path

import pandas as pd

folder = sorted(Path("../data/static/gdynia").iterdir())[-1]
print(folder)
```

```python
stop_times = pd.read_csv(folder / "stop_times.txt")
counts = stop_times["stop_id"].value_counts()
top = counts.head(40)
top
```

```python
stops = pd.read_csv(folder / "stops.txt")
names = stops.set_index("stop_id")["stop_name"]
top.to_frame("buses").join(names)
```

```python
import requests

r = requests.get("https://api.zdiz.gdynia.pl/pt/stops", timeout=10)
api_ids = {s["stopId"] for s in r.json()}
print([s for s in top.index if s not in api_ids])
```
