## **Downloading static GTFS general function**

```python
from datetime import date
from pathlib import Path

import requests


def download_one_gtfs(origin: str, url: str) -> None:

    today = date.today()
    gtfs_date = today.isoformat()
    gtfs_origin_file = Path("data/static") / origin / gtfs_date / "gtfs.zip"
    if gtfs_origin_file.exists() and gtfs_origin_file.stat().st_size > 0:
        print(f"File {origin}_gtfs already exists")
    else:
        static_gtfs_origin = url
        response_gtfs_origin = requests.get(static_gtfs_origin, timeout=20)
        response_gtfs_origin.raise_for_status()
        print(response_gtfs_origin)
        gtfs_origin_file.parent.mkdir(parents=True, exist_ok=True)
        gtfs_origin_file.write_bytes(response_gtfs_origin.content)
```

## **Downloading GTFS for Gdansk**

```python
gdansk_url = "https://ckan.multimediagdansk.pl/dataset/c24aa637-3619-4dc2-a171-a23eec8f2172/resource/30e783e4-2bec-4a7d-bb22-ee3e3b26ca96/download/gtfsgoogle.zip"
origin = "gdansk"
download_one_gtfs(origin, gdansk_url)
```

## **Downloading GTFS for SKM, Gdansk and Gdynia in one call**

```python
def download_gtfs(base_dir=Path("data/static")):
    sources: dict[str, str] = {
        "gdansk": "https://ckan.multimediagdansk.pl/dataset/c24aa637-3619-4dc2-a171-a23eec8f2172/resource/30e783e4-2bec-4a7d-bb22-ee3e3b26ca96/download/gtfsgoogle.zip",
        "gdynia": "https://api.zdiz.gdynia.pl/pt/gtfs.zip",
        "skm": "https://www.skm.pkp.pl/gtfs-mi-kpd.zip",
    }

    today = date.today()
    file_date = today.isoformat()
    for origin, url in sources.items():
        file_name = Path(base_dir) / origin / file_date / "gtfs.zip"
        if file_name.exists() and file_name.stat().st_size > 0:
            print(f"File {file_name} already exists")
        else:
            try:
                response_gtfs = requests.get(url, timeout=20)
                response_gtfs.raise_for_status()
                file_name.parent.mkdir(parents=True, exist_ok=True)
                file_name.write_bytes(response_gtfs.content)
            except requests.exceptions.RequestException as error:
                print(f"{origin}: URL failed ({error}), skipping")

    if __name__ == "__main__":
        download_gtfs()
```
